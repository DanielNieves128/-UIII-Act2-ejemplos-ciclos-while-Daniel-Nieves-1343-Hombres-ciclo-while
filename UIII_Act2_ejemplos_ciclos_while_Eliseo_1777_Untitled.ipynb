```python
# EJEMPLO 1: Contar del 1 al 5

contador = 1

while contador <= 5:
    print(contador)
    contador += 1
```

```python
# EJEMPLO 2: Mostrar números pares

numero = 2

while numero <= 10:
    print(numero)
    numero += 2
```

```python
# EJEMPLO 3: Pedir contraseña correcta

clave = ""

while clave != "python":
    clave = input("Ingresa la contraseña: ")

print("Acceso permitido")
```

```python
# EJEMPLO 4: Suma de números del 1 al 5

contador = 1
suma = 0

while contador <= 5:
    suma = suma + contador
    contador += 1

print("La suma es:", suma)
```

```python
# EJEMPLO 5: Validar número positivo

numero = -1

while numero <= 0:
    numero = int(input("Ingresa un número positivo: "))

print("Número válido")
```

```python
# EJEMPLO 6: Tabla de multiplicar

numero = 3
contador = 1

while contador <= 10:
    print(numero, "x", contador, "=", numero * contador)
    contador += 1
```
